# Basics: Funktionen
**Funktionen** sind der grundlegender Bestandteil eines Programms. Eine Funktion ist eine festgegte Abfolge von **Operationen**, die mit einer Reihe von Variablen, den **Argumenten**, ausgeführt wird. Die Argumente werden durch **runde Klammern** `()` festgelegt, die allgemein das Merkmal von Funktionen sind. Ein Beispiel dafür ist die `print()` Funktion, die alles innerhalb der Klammern ausgibt.

````python
a = 1
b = 2
print("Die Zahl", 1, "ist kleiner als", 2)
````

In Python gibt es eine Reihe von **Built-In Funktionen**, die wesentliche Funktionalitäten zur Verfügung stellen. In der Regel ist es aber notwendig, zusätzlich eigene Funktionen zu nutzen.


# Basics: Funktionen definieren
Der Beginn einer Definition für eine Funktion wird durch das **def** Keyword signalisiert. Die allgemeine Syntax ist

```python 
    def name(argument1, argument2, ...): 
        #Code
        return output1, output2, ...
```

````python
def exponent(x, power):
    """
    Bildet den Exponenten
    Input:
    x: Float
    Output:
    y: Float
    """
    y = x ** power
    return y

a = 5
b = exponent(a, 2)
print(a, b)
````

# Funktionen verzweigen
Funktionen müssen nicht immer gleich funktionieren - in manchen Fällen kann es nützlich sein Ausnahmen zu definieren

### Beispiel: Division durch 0

````python
def division(a, b):
    """
    Dividiere a / b
    Input:
    a: Integer / Float
    b: Integer / Float
    """
    # Prüfe ob b sehr klein ist
    if abs(b) < 10e-12:
        # Return False
        return False
    # Ansonsten: Normale Funktion
    return a / b

div1 = division(10, 2.5)
div2 = division(10, 0)

print(div1, div2)
````

# Numpy: Funktionen und Arrays

# Numpy: Achtung bei Array Kopien!


````python
import numpy as np
import copy
a = np.array([1, 1, 1])
b = a 
b[0] = 2
print(a, b)
````

**Grund:** Python speichert  alles als Objekte. Mit `b = a` enthält die variable `b` die Referenz zu dem Objekt der Variable `a`, aber kein neues Objekt wird erzeugt.

**Workaround:**
```python 
import numpy as np 
import copy
b = copy.deepcopy(a)
```
